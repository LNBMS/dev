```python
import pandas as pd
import numpy as np
```

create a sample dataset

```python
data = {
    "Name": ["Arun", "Priya", "Rahul", "Sneha", "Kiran"],
    "Department": ["CSE", "ECE", "CSE", "IT", "ECE"],
    "Age": [20, 21, 22, 20, 23],
    "Marks": [85, 92, 78, 88, 95],
    "City": ["Chennai", "Bengaluru", "Hyderabad",
             "Chennai", "Bengaluru"]
}

df = pd.DataFrame(data)
print(df)
```

save dataset to csv

```python
df.to_csv("students.csv", index=False)
```

**to_csv()** saves the DataFrame as a CSV file.



**index=False** prevents Pandas from saving an extra row-number column.

now load the csv file

```python
df_csv = pd.read_csv("students.csv")
print(df_csv)
```

Task:Change rahul marks from 78 to 90  , save the file again, and reload it.

```python
df.loc[df["Name"] == "Rahul", "Marks"] = 90

#save the updated data to csv
df.to_csv("students.csv", index=False)

print("Updated data saved successfully!")
df_reloaded = pd.read_csv("students.csv")

print(df_reloaded)
#verify the updated marks in csv
print(df_reloaded[df_reloaded["Name"] == "Rahul"])
```

**JSON FILE**
JSON is commonly used for exchanging structured data between web applications and APIs.


Load JSON file

```python
df.to_json("students.json", orient="records", indent=2)

df_json = pd.read_json("students.json")
print(df_json)
```

**df.to_json("students.json", orient="records", indent=2)**
df	The DataFrame containing your student data

.to_json()	Converts the DataFrame into JSON format

"students.json"	The name of the output file

orient="records"	Stores each row as a separate JSON object
Records format — one object per row
[
  {"Name": "Arun", "Department":CSE ,"Age": 20,"Marks": 85,"City":Chennai},

  {"Name": "Priya", "Department":ECE ,"Age": 21,"Marks": 92,"City":Bengaluru}
]


Without orient="records", Pandas uses its default orient="columns" format, which arranges the JSON differently.

indent=2	Adds indentation of 2 spaces to make the file readable

Without indentation:


[{"Name":"Arun","Department":"CSE",
    "Age": 20,"Marks":85,"City":"Chennai"},
    {"Name":"Priya","Department":"ECE",
    "Age": 21,"Marks":92,"City":"Benagluru"}]


With indent=2:


[
  {
    "Name": "Arun",
    "Department":"CSE",
    "Age": 20,
    "Marks": 85,
    "City":"Chennai"
  },
  {
    "Name": "Priya",
    "Department":"ECE",
    "Age": 21,
    "Marks": 92,

  }
]
