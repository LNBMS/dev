```python
import requests
import pandas as pd
import matplotlib.pyplot as plt
```

define api endpoint

```python
url = "https://jsonplaceholder.typicode.com/posts"

print(url)
```

send a get request

```python
response = requests.get(url, timeout=15)

print(response.status_code)
```

Read the JSON response

```python
posts = response.json()

print(type(posts))
print(len(posts))
print(posts[0])
```

```python
print(posts[0])
```

create a dataframe

```python
df = pd.DataFrame(posts)

print(df.head())
```
