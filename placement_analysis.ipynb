```python
import kagglehub
import os
import pandas as pd

# Download dataset
path = kagglehub.dataset_download("navinpatidar/indian-student-placement-data")

# Find CSV file
csv_file = [f for f in os.listdir(path) if f.endswith(".csv")][0]

# Load dataset
df = pd.read_csv(os.path.join(path, csv_file))

# Display first 5 records
df.head()
```

```python

#rename col
print("Cols:",df.columns)
# The following rename operations are commented out because the source columns 'CGPA', 'PlacementStatus', '10th percentage', '12th percentage', 'Degree Percentage' do not exist in the DataFrame.
# Please verify the correct column names from `df.columns` above and modify the rename operations accordingly.
df=df.rename(columns={"CGPA": "CGPA","PlacementStatus":"Placement_Status"})
#rename several col
df=df.rename(columns={
    "10th percentage":"Percentage_10th",
    "12th percentage":"Percentage_12th",
    "Degree Percentage":"Degree_Percentage"
 })

# The following lines are commented out as they refer to non-existent columns (CGPA, Placement_Status, Degree_Percentage, Specialization).
# To fix this, please use existing column names from the `df.columns` output.
df.columns
#select single col
df["CGPA"]
#Select multiple col
df[ [
     "CGPA",
    "Degree_Percentage",
    "Placement_Status"
 ]
    ]
#using loc
df.loc[:,["CGPA","Degree_Percentage"]]
#specify rows
df.loc[0:10,["CGPA","Degree_Percentage"]]
#using iloc
df.iloc[0:10,0:5]
#specific row
df.iloc[5]
# students having cgpa greater than 8
df[df["CGPA"]>8]
#Students with CGPA >8 and degree percentage >70
df[
    (df["CGPA"]>8) &
    (df["Degree_Percentage"]>70)
 ]
#using isin()
df[
    df["Specialization"].isin(
       ["Computer Science","Information Technology"]
   )
 ]
#using between()
#Students whose CGPA is between 7 and 8.5
df[
    df["CGPA"].between(7,8.5)
 ]
#sorting -highest cgpa first
df.sort_values(
   by="CGPA",
    ascending=False
 ).head(10)
#Sorting using multiple col
df.sort_values(
    by=["Specialization","CGPA"],
    ascending=[True, False]
 )
#value_counts()
df["Placement_Status"].value_counts()
#display unique value_counts
df["Placement_Status"].unique()
#display no of unique values
df["Placement_Status"].nunique()
#percentage
#df["Placement"]
```
