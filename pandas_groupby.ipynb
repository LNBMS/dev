Dataframe can be divided into smaller groups based on the value
in these columns.This can be done by using -groupby


** group data by a single column and then calculate the aggregates.**

```python
import pandas as pd

# create a dictionary containing the data
data = {'Category': ['Jewllery', 'Clothing', 'Jewllery', 'Clothing'],
        'Sales': [2000, 1500, 800, 300]}

# create a DataFrame using the data dictionary
df = pd.DataFrame(data)
print(df)

# group the DataFrame by the Category column and
# calculate the sum of Sales for each category
grouped = df.groupby('Category')['Sales'].sum()

# print the grouped data
print(grouped)
```

**group by multiple col in pandas**

```python
import pandas as pd

# create a DataFrame with student data
data = {
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male'],
    'Grade': ['A', 'B', 'A', 'A', 'B'],
    'Score': [90, 85, 92, 88, 78]
}

df = pd.DataFrame(data)

# define the aggregate functions to be applied to the Score column
agg_functions = {
    # calculate both mean and maximum of the Score column
    'Score': ['mean', 'max']
}

# group the DataFrame by Gender and Grade, then apply the aggregate functions
grouped = df.groupby(['Gender', 'Grade']).aggregate(agg_functions)

# print the resulting grouped DataFrame
print(grouped)
```

**group with categorical data**

```python
import pandas as pd

# sample data
data = {'Category': ['Apple', 'Banana', 'Apple', 'Banana', 'Apple', 'Banana'],
        'Sales': [100, 150, 200, 50, 300, 120]}

df = pd.DataFrame(data)

# convert Category column to categorical type
df['Category'] = pd.Categorical(df['Category'])

# group by Category  and calculate the total sales
grouped = df.groupby('Category')['Sales'].sum()

print(grouped)
```

```python
import pandas as pd

data = {
    'Department': ['IT', 'HR', 'IT', 'Sales', 'HR',
                   'Sales', 'IT', 'HR', 'Sales', 'IT'],
    'Employee': ['Amit', 'Priya', 'Rahul', 'Sneha', 'Kiran',
                 'Arjun', 'Neha', 'Ravi', 'Pooja', 'Anil'],
    'Salary': [60000, 45000, 75000, 50000, 48000,
               55000, 80000, 52000, 58000, 70000],
    'Age': [25, 30, 28, 26, 32, 29, 27, 35, 31, 24],
    'Experience': [2, 5, 4, 3, 6, 4, 5, 8, 7, 2],
    'Gender': ['M', 'F', 'M', 'F', 'M',
               'M', 'F', 'M', 'F', 'M']
}

df = pd.DataFrame(data)
print(df)
```

Find the total salary paid by each department.

```python
grouped = df.groupby('Department')['Salary'].sum()
print(grouped)
```

Average salary by department

```python
grouped = df.groupby('Department')['Salary'].mean()
print(grouped)
```

Employee count by department

```python
grouped = df.groupby('Department')['Employee'].count()
print(grouped)
```

Highest sal by dept

```python
grouped = df.groupby('Department')['Salary'].max()
print(grouped)
```

Minimum employee age by department -Find the youngest employee's age in each department.

```python
grouped = df.groupby('Department')['Age'].min()
print(grouped)
```

minimum age by department along with the employee name

```python
idx = df.groupby('Department')['Age'].idxmin()

result = df.loc[idx, ['Department', 'Employee', 'Age']]
print(result)
```

**group by multiple columns**

Average salary by department and gender

```python
grouped = df.groupby(['Department', 'Gender'])['Salary'].mean()   #group by department and gender
print(grouped)
```

Employee count by department and gender

```python
grouped = df.groupby(['Department', 'Gender']).size()
print(grouped)
```

Total experience by department and gender

```python
grouped = df.groupby(['Department', 'Gender'])['Experience'].sum()
print(grouped)
```

Maximum salary by department and gender

```python
grouped = df.groupby(['Department', 'Gender'])['Salary'].max()
print(grouped)
```

Multiple aggregation functions

Minimum, maximum, and average salary by department

```python
grouped = df.groupby('Department')['Salary'].agg(
    ['min', 'max', 'mean']
)
print(grouped)
```

grouped = df.groupby('Department')['Salary'].agg(
    ['min', 'max', 'mean']
)
print(grouped)

```python
grouped = df.groupby('Department').agg({
    'Age': 'mean',
    'Experience': 'mean'
})
print(grouped)
```

Different aggregations on different columns

```python
grouped = df.groupby('Department').agg({
    'Salary': 'sum',
    'Age': 'mean',
    'Experience': 'max'
})
print(grouped)
```

Sort departments by average salary

```python
grouped = df.groupby('Department')['Salary'].mean()

grouped = grouped.sort_values(ascending=False)
print(grouped)
```

 Employees earning above their department average

```python
avg_salary = df.groupby('Department')['Salary'].transform('mean')

result = df[df['Salary'] > avg_salary]
print(result)
```

Departments with average salary above ₹55,000

```python
grouped = df.groupby('Department')['Salary'].mean()

result = grouped[grouped > 55000]
print(result)
```

Count employees with at least 5 years of experience

```python
experienced = df[df['Experience'] >= 5]

result = experienced.groupby('Department').size()
print(result)
```
