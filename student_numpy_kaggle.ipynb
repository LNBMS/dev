```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("lainguyn123/student-performance-factors")

print("Path to dataset files:", path)
```

```python
import numpy as np $0

data = np.genfromtxt(
    "/kaggle/input/student-performance-factors/StudentPerformanceFactors.csv",
    delimiter=",",
    skip_header=1,
    usecols=(0,1,5,6,8,19),
    dtype=np.int32
)

print(data.shape)
print(data)
```

**delimiter=","**
A CSV means Comma Separated Values.
---
**skip_header=1**
in csv,first row contains col names not values.If you don't skip it:NumPy tries to convert "Hours_Studied" into an integer and gives an error or nan
---
**usecols=(0,1,5,6,8,19)**
dataset is having 20 columns,in that we are choosing which are required for our project

0 Hours_Studied

1  Attendance

5 Previous_Scores

6 Tutoring_Sessions

8 Sleep_Hours

19 Exam_Score

---
**dtype=np.int32**
every number stored as a 32 bit integer
---

output(6607,6)-

6607- no of students
6- no of features

```python
#boolean masking
#find students who's score is <35
exam = data[:,5]
failed = exam < 35
failed_students = data[failed]
print(failed_students)
```

```python
# Attendance above 90
high_att = data[data[:,1] > 90]
print("attendance >90",high_att)
# Sleep less than 5 hours
poor_sleep = data[data[:,4] < 5]
print("sleep less than 5 hrs:",poor_sleep)
# Tutoring more than 3 sessions
extra = data[data[:,3] > 3]
print("tutoring more than 3 sessions:",extra)
```

```python
#statistics
print("mean of scores:",np.mean(data[:,5]))   #class average

print("median of scores:",np.median(data[:,5]))  #class middle performer

print("std of scores:",np.std(data[:,5])) #difficulty level

print("max of scores:",np.max(data[:,5])) #topper

print("min of scores:",np.min(data[:,5])) #lowest score
```
